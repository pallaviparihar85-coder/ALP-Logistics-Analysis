```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
```

```python
df = pd.read_csv("APL_Logistics.csv", encoding="latin1")
```

```python
# Compute Total Revenue and Profit

total_revenue = df["Sales"].sum()
total_profit = df["Order Profit Per Order"].sum()

print("Total Revenue :", total_revenue)
print("Total Profit  :", total_profit)
```

```python
# Compare Revenue vs Profit by Market

market_analysis = (
    df.groupby("Market")[["Sales", "Order Profit Per Order"]]
    .sum()
    .reset_index()
)

market_analysis.columns = [
    "Market",
    "Total Revenue",
    "Total Profit"
]

market_analysis
```

```python
plt.figure(figsize=(12,6))

x = np.arange(len(market_analysis))

plt.bar(x-0.2,
        market_analysis["Total Revenue"],
        width=0.4,
        label="Revenue")

plt.bar(x+0.2,
        market_analysis["Total Profit"],
        width=0.4,
        label="Profit")

plt.xticks(x,
           market_analysis["Market"],
           rotation=45)

plt.title("Revenue vs Profit by Market")
plt.xlabel("Market")
plt.ylabel("Amount")
plt.legend()

plt.show()
```

```python
profit_pattern = (
    df.groupby("Market")["Order Profit Per Order"]
    .sum()
    .reset_index()
)

profit_pattern.columns = [
    "Market",
    "Total Profit"
]

profit_pattern = profit_pattern.sort_values(
    by="Total Profit",
    ascending=False
)

profit_pattern
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    profit_pattern["Market"],
    profit_pattern["Total Profit"]
)

plt.title("Profit Concentration by Market")
plt.xlabel("Market")
plt.ylabel("Total Profit")

plt.xticks(rotation=45)

plt.show()
```

```python
# Profit by Product Name

product_profit = (
    df.groupby("Product Name")["Order Profit Per Order"]
    .sum()
    .reset_index()
    .sort_values(by="Order Profit Per Order", ascending=False)
)

product_profit.head(10)
```

```python
top_product_profit = product_profit.head(10)

plt.figure(figsize=(12,6))

plt.bar(
    top_product_profit["Product Name"],
    top_product_profit["Order Profit Per Order"]
)

plt.title("Top 10 Products by Profit")
plt.xlabel("Product Name")
plt.ylabel("Total Profit")

plt.xticks(rotation=90)

plt.show()
```

```python
# Profit Margin by Product Name

product_margin = (
    df.groupby("Product Name")["Order Item Profit Ratio"]
    .mean()
    .reset_index()
    .sort_values(by="Order Item Profit Ratio", ascending=False)
)

product_margin.head(10)
```

```python
top_product_margin = product_margin.head(10)

plt.figure(figsize=(12,6))

plt.bar(
    top_product_margin["Product Name"],
    top_product_margin["Order Item Profit Ratio"]
)

plt.title("Top 10 Products by Profit Margin")
plt.xlabel("Product Name")
plt.ylabel("Profit Margin")

plt.xticks(rotation=90)

plt.show()
```

```python
# Profit by Category

category_profit = (
    df.groupby("Category Name")["Order Profit Per Order"]
    .sum()
    .reset_index()
    .sort_values(by="Order Profit Per Order", ascending=False)
)

category_profit
```

```python
plt.figure(figsize=(14,6))

plt.bar(
    category_profit["Category Name"],
    category_profit["Order Profit Per Order"]
)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.xticks(rotation=45)

plt.show()
```

```python
# Profit Margin by Category

category_margin = (
    df.groupby("Category Name")["Order Item Profit Ratio"]
    .mean()
    .reset_index()
    .sort_values(by="Order Item Profit Ratio", ascending=False)
)

category_margin
```

```python
plt.figure(figsize=(16,5))

plt.bar(
    category_margin["Category Name"],
    category_margin["Order Item Profit Ratio"]
)

plt.title("Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin")

plt.xticks(rotation=45)

plt.show()
```

```python
high_revenue_low_margin = (
    df.groupby("Product Name")
    .agg({
        "Sales":"sum",
        "Order Item Profit Ratio":"mean"
    })
    .reset_index()
)

high_revenue_low_margin = high_revenue_low_margin.sort_values(
    by="Sales",
    ascending=False
)

high_revenue_low_margin.head(10)
```

```python
top_products = high_revenue_low_margin.head(10)

plt.figure(figsize=(12,6))

plt.scatter(
    top_products["Sales"],
    top_products["Order Item Profit Ratio"]
)

for i in range(len(top_products)):
    plt.text(
        top_products["Sales"].iloc[i],
        top_products["Order Item Profit Ratio"].iloc[i],
        top_products["Product Name"].iloc[i],
        fontsize=7
    )

plt.title("High Revenue vs Low Margin Products")
plt.xlabel("Sales")
plt.ylabel("Profit Margin")

plt.show()
```

```python
# Aggregate Sales and Profit by Customer Id

customer_analysis = (
    df.groupby("Customer Id")[["Sales", "Order Profit Per Order"]]
    .sum()
    .reset_index()
)

customer_analysis.columns = [
    "Customer Id",
    "Total Sales",
    "Total Profit"
]

customer_analysis.head(10)
```

```python
top_customers = customer_analysis.sort_values(
    by="Total Sales",
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

x = np.arange(len(top_customers))

plt.bar(
    x-0.2,
    top_customers["Total Sales"],
    width=0.4,
    label="Sales"
)

plt.bar(
    x+0.2,
    top_customers["Total Profit"],
    width=0.4,
    label="Profit"
)

plt.xticks(
    x,
    top_customers["Customer Id"].astype(str),
    rotation=90
)

plt.title("Top 10 Customers: Sales vs Profit")
plt.xlabel("Customer Id")
plt.ylabel("Amount")
plt.legend()

plt.show()
```

```python
# High-Value Customers

high_value_customers = customer_analysis.sort_values(
    by="Total Sales",
    ascending=False
).head(10)

high_value_customers
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    high_value_customers["Customer Id"].astype(str),
    high_value_customers["Total Sales"]
)

plt.title("Top 10 High-Value Customers")
plt.xlabel("Total Sales")
plt.ylabel("Customer Id")

plt.show()
```

```python
# Loss-Making Customers

loss_customers = customer_analysis[
    customer_analysis["Total Profit"] < 0
]

loss_customers
```

```python
loss_top10 = loss_customers.sort_values(
    by="Total Profit"
).head(10)

plt.figure(figsize=(10,6))

plt.barh(
    loss_top10["Customer Id"].astype(str),
    loss_top10["Total Profit"]
)

plt.title("Top 10 Loss-Making Customers")
plt.xlabel("Total Profit")
plt.ylabel("Customer Id")

plt.show()
```

```python
# Segment Customers by Value Tiers

customer_analysis["Customer Tier"] = pd.qcut(
    customer_analysis["Total Sales"],
    q=3,
    labels=["Low Value", "Medium Value", "High Value"]
)

customer_analysis[
    ["Customer Id", "Total Sales", "Customer Tier"]
].head(10)
```

```python
tier_count = customer_analysis["Customer Tier"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    tier_count,
    labels=tier_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Value Tiers")

plt.show()
```

```python
# Compare Profit Margins with and without Discounts

discount_analysis = df.copy()

discount_analysis["Discount Status"] = discount_analysis[
    "Order Item Discount Rate"
].apply(lambda x: "Without Discount" if x == 0 else "With Discount")

discount_summary = (
    discount_analysis.groupby("Discount Status")
    .agg({
        "Sales":"sum",
        "Order Profit Per Order":"sum",
        "Order Item Profit Ratio":"mean"
    })
    .reset_index()
)

discount_summary.columns = [
    "Discount Status",
    "Total Sales",
    "Total Profit",
    "Profit Margin"
]

discount_summary
```

```python
plt.figure(figsize=(6,5))

plt.bar(
    discount_summary["Discount Status"],
    discount_summary["Profit Margin"]
)

plt.title("Profit Margin: With vs Without Discount")
plt.xlabel("Discount Status")
plt.ylabel("Profit Margin")

plt.show()
```

```python
# Identify Discount Thresholds Causing Margin Erosion

margin_erosion = (
    df.groupby("Order Item Discount Rate")
    .agg({
        "Order Item Profit Ratio": "mean",
        "Order Profit Per Order": "mean"
    })
    .reset_index()
)

margin_erosion.columns = [
    "Discount Rate",
    "Average Profit Margin",
    "Average Profit"
]

margin_erosion
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    margin_erosion["Discount Rate"],
    margin_erosion["Average Profit Margin"],
    marker="o"
)

plt.title("Margin Erosion by Discount Rate")
plt.xlabel("Discount Rate")
plt.ylabel("Average Profit Margin")

plt.grid(True)

plt.show()
```

# Business Insights

- Total revenue is significantly higher than total profit, indicating opportunities to improve overall profitability.
- Some products generate high sales but relatively lower profit margins, suggesting the need for better pricing strategies.
- High-value customers contribute a major portion of total revenue and should be prioritized through retention programs.
- Higher discount rates are associated with lower profit margins, highlighting the impact of excessive discounting.
- Profitability varies across markets, regions, and countries, indicating that some locations perform better than others.
- A few categories and customers generate lower or negative profits, requiring focused business attention.

# Recommendations

- Reduce excessive discounts on products with already low profit margins.
- Focus marketing and sales efforts on high-profit products and profitable markets.
- Review pricing strategies for high-revenue but low-margin products.
- Strengthen customer loyalty programs to retain high-value customers.
- Monitor loss-making categories and customers to improve profitability.
- Optimize regional operations by investing more in high-performing markets and improving weaker regions.
